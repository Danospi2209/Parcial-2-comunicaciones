# Analisis de datos y trafico
Este cuaderno (1) consulta PostgreSQL y (2) procesa el log de acceso de Nginx hacia Joomla.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from sqlalchemy import create_engine

url = (f"postgresql+psycopg2://{os.environ['POSTGRES_USER']}:{os.environ['POSTGRES_PASSWORD']}"
       f"@{os.environ['POSTGRES_HOST']}:5432/{os.environ['POSTGRES_DB']}")
engine = create_engine(url)
print('Conexion OK:', pd.read_sql('SELECT version()', engine).iloc[0, 0])
```

## 1. Tablas de la base de datos

```python
pd.read_sql("""SELECT table_name FROM information_schema.tables
                WHERE table_schema='public' ORDER BY 1""", engine)
```

## 2. Metricas de PostgreSQL en el tiempo

```python
m = pd.read_sql('SELECT * FROM metricas ORDER BY ts', engine)
m['transacciones'] = m['xact_commit'].diff().clip(lower=0)
fig, ax = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
ax[0].plot(m['ts'], m['conexiones']); ax[0].set_ylabel('Conexiones')
ax[1].plot(m['ts'], m['transacciones']); ax[1].set_ylabel('Transacciones')
plt.xlabel('Tiempo'); plt.tight_layout(); plt.show()
```

## 3. Procesamiento del log de Nginx (trafico hacia Joomla)
Genera trafico navegando en `http://localhost/` antes de ejecutar.

```python
import re
pat = re.compile(r'(?P<ip>\S+) - \S+ \[(?P<fecha>[^\]]+)\] "(?P<metodo>\S+) (?P<ruta>\S+) [^"]*" (?P<status>\d{3}) (?P<bytes>\d+)')
rows = []
with open('/home/jovyan/logs/access.log') as f:
    for line in f:
        m_ = pat.match(line)
        if m_: rows.append(m_.groupdict())
df = pd.DataFrame(rows)
print(len(df), 'peticiones')
df.head()
```

```python
if not df.empty:
    fig, ax = plt.subplots(1, 2, figsize=(12, 4))
    df['status'].value_counts().sort_index().plot.bar(ax=ax[0], title='Peticiones por codigo HTTP')
    df['ip'].value_counts().head(10).plot.barh(ax=ax[1], title='Top 10 IPs')
    plt.tight_layout(); plt.show()
else:
    print('Aun no hay peticiones registradas.')
```
